## Question 1: Which genres receive the highest player ratings?


```python
# Sub-question: Which game genres have the highest average player ratings?
def genre_rating_summary(games, n=10):
    # Genres may be separated by commas; split them into individual genre rows.
    genre_rows = games[["genres", "pct_pos_total"]].dropna().copy()
    genre_rows["genres"] = genre_rows["genres"].astype(str).str.split(",")
    genre_rows = genre_rows.explode("genres")
    genre_rows["genres"] = genre_rows["genres"].str.strip()
    genre_rows = genre_rows[genre_rows["genres"] != ""]
    average = genre_rows.groupby("genres")["pct_pos_total"].mean().round(1)
    return average.sort_values(ascending=False).head(n)

my_function = genre_rating_summary
my_result = genre_rating_summary(games, 10)
print(my_result)
```
